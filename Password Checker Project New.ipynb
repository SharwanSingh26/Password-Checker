```python
import time
def passwordchecker(passs):
    special_chars = "!@#$%^&*()"
    points =0
    print("Checking Password Please wait...")
    time.sleep(3)
    if passs =='':
        print('Enter Some Text Value')
    else:
        if len(passs) >=8:
            points+=1
            if any(char in special_chars for char in passs):
                points+=1
            else:
                print("No special character")
            if any(char.isupper() for char in passs):
                points+=1
            else:
                print('Password Have One UpperCase characters')
            if any(char.isdigit() for char in passs):
                points+=1
            else:
                print('Password Have One Digit')

            if points <1:
                return "Weak"
               
            elif points ==2 or points == 3:
                return "Medium"
                
            else:
                return "Strong"
                  
        else:
            print('Password Have atleast 8 characters')


#I Use Loop for Give Limited attempts To Check Password 
for i in range(4):
    passs = input("Enter your password: ")
    strength = passwordchecker(passs)
    
    if strength == "Weak":
        print(f"Try again. Attempts left: {3 - i}")
        print('Your Password Is Very Weak')
    elif  strength == "Medium":
        print(f"Try again. Attempts left: {3 - i}")
        print('Your Password Is Medium')
    elif strength == "Strong":
        print('Password Is Strong')
        break
    else:
        print(f"Try again. Attempts left: {3 - i}")

else:
    print("Too many attempts. Try later")
```
